# Constrained Optimization

Find the maximum of $f(x_1, x_2)$ on the constraint set $\mathcal{X} = \{(x_1, x_2) : g(x_1, x_2) = \text{level}\}$ by reading the plot.

At a constrained maximum, the point lies on $\mathcal{X}$ and $\nabla f$ is parallel to $\nabla g$:

$$\nabla f(x) = \lambda \nabla g(x)$$
